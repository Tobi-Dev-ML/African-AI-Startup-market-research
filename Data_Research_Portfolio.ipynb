## 4. Research Scope & Inclusion Criteria

### Research Population

The research focuses on African AI-focused companies that were founded in Africa or substantially operate in African markets.

### Inclusion Criteria

A company may be included when:

- AI is a meaningful component of its product, service, or core technology.
- The company has a publicly identifiable product, service, or business activity.
- The company has a documented connection to an African country or African market.
- Sufficient public information is available to verify the company's identity and key details.
- Relevant company or funding information can be supported by reliable public sources.

### Exclusion Criteria

The following are generally excluded:

- Companies with no meaningful AI component.
- AI companies with no verifiable connection to African markets.
- Individuals, projects, research papers, or academic laboratories without a company-level business.
- Companies for which basic identity or operating information cannot be reliably verified.
- Duplicate entries representing the same company.

### Research Principle

The dataset prioritizes information quality and source verification over the number of companies collected.

Where information is unavailable or funding has not been publicly disclosed, the information will not be estimated. Missing or undisclosed information will be recorded appropriately.

## 5. Research Methodology

The research follows a structured company-by-company process:

1. **Company Identification**  
   Identify potentially relevant African AI-focused companies through company websites, startup databases, technology publications, funding reports, industry research, and other reputable sources.

2. **Eligibility Screening**  
   Determine whether each company meets the project's inclusion criteria.

3. **Company Research**  
   Collect information on the company's location, founding year, AI focus, industry, product or service, target market, and business model.

4. **Funding Research**  
   Search for publicly disclosed funding information, including funding stage, total disclosed funding, latest funding date, latest round, and lead investor.

5. **Source Verification**  
   Record sources supporting company information and funding information separately.

6. **Conflict Checking**  
   Investigate conflicting information such as different founding years, headquarters, or funding amounts before recording the final value.

7. **Data Standardization**  
   Apply consistent formats and terminology while preserving the meaning of the original information.

8. **Quality Control**  
   Check for duplicate companies, missing critical information, inconsistent values, unsupported claims, and unresolved research issues.

9. **Analysis**  
   Analyze the completed dataset to identify patterns in geography, AI applications, industries, funding stages, and other research dimensions.

10. **Final Validation and Export**  
    Validate the final dataset and prepare the research files and supporting reports for delivery.

## 6. Source & Verification Framework

### Source Priority

Information will be collected using a source hierarchy. Primary sources will be preferred whenever available.

**Tier 1 — Primary Sources**
- Official company website
- Official company newsroom or announcements
- Official company LinkedIn page
- Investor or funding announcement
- Official regulatory or corporate filing, where applicable

**Tier 2 — Reputable Secondary Sources**
- Established technology and business publications
- Reputable startup and venture-capital databases
- Industry research reports
- Recognized African technology and business publications

**Tier 3 — Discovery Sources**
- Startup directories
- Search results
- Aggregator websites
- Social media posts or other sources used primarily to identify potential companies or leads

Tier 3 sources may help identify companies or information, but important claims should be verified using stronger sources where possible.

### Verification Approach

Company information and funding information will be verified separately.

**Company information** may include:
- Company name
- Country
- City
- Founded year
- AI focus
- Industry
- Product or service
- Target market
- Business model

**Funding information** may include:
- Funding stage
- Total publicly disclosed funding
- Latest funding date
- Latest funding round
- Lead investor

A source will be recorded in the appropriate source column so that major claims in the dataset can be traced back to supporting evidence.

### Handling Conflicting Information

When reliable sources provide different information:

1. Identify the conflicting values.
2. Compare the dates and context of the sources.
3. Give greater weight to recent primary sources where appropriate.
4. Check whether the difference is caused by different definitions or reporting periods.
5. Record the selected value only when there is sufficient evidence.
6. Document unresolved conflicts in `Research_Notes`.

### Funding Data Rule

Only publicly disclosed funding amounts will be recorded.

Funding amounts will not be estimated from:
- Company valuation
- Investor ownership
- Unverified database figures
- Rumours
- Speculation
- Undisclosed deal terms

When a reliable funding amount cannot be established, the amount will remain blank and the situation will be documented in the relevant research fields.

### Verification Status

Each company will receive one of three verification statuses:

- **Verified** — Key company information and relevant funding information have sufficient supporting evidence.
- **Partially Verified** — Core company information is supported, but one or more important fields require additional verification.
- **Needs Review** — Important information remains uncertain, conflicting, or insufficiently supported.

### 12.2 Date Standardization

## 15. Final Data Cleaning & Standardization

The 50-company research dataset is now reviewed as a complete dataset.

The final cleaning process standardizes text fields, categorical values, dates, funding values, missing information, and company names while preserving the underlying research findings.

No missing information will be replaced with assumptions.

```python
# Create a separate working copy so the original staging dataset
# remains available for comparison if required.

df_final = research_staging.copy()

print("Rows:", df_final.shape[0])
print("Columns:", df_final.shape[1])
```

```python
# Confirm that the final dataset still follows the established
# 20-column structure.

print("Column structure valid:", list(df_final.columns) == expected_columns)
```

```python
# Remove accidental leading/trailing spaces from text fields
# without changing the actual research content.

text_columns = df_final.select_dtypes(include="object").columns

for column in text_columns:
    df_final[column] = df_final[column].apply(
        lambda value: value.strip() if isinstance(value, str) else value
    )

print("Text fields standardized.")
```

```python
# Convert empty strings and whitespace-only values to missing values.
# This creates one consistent representation of unavailable information.

df_final = df_final.replace(r"^\s*$", pd.NA, regex=True)

print("Blank values standardized as missing values.")
```

```python
# Review missing values across the completed 50-company dataset.

missing_summary = (
    df_final.isna()
    .sum()
    .sort_values(ascending=False)
)

missing_summary[missing_summary > 0]
```

```python
# Confirm that the final cleaning process has not introduced
# or retained duplicate company records.

duplicate_companies = df_final[
    df_final.duplicated(subset=["Company_Name"], keep=False)
]

print("Duplicate companies:", len(duplicate_companies))
```

```python
# Verify that categorical fields still contain only approved values.

print(
    "AI Focus valid:",
    set(df_final["AI_Focus"].dropna()).issubset(approved_ai_focus)
)

print(
    "Industry valid:",
    set(df_final["Industry"].dropna()).issubset(approved_industries)
)

print(
    "Funding Stage valid:",
    set(df_final["Funding_Stage"].dropna()).issubset(
        approved_funding_stages
    )
)

print(
    "Verification Status valid:",
    set(df_final["Verification_Status"].dropna()).issubset(
        approved_verification_status
    )
)
```

```python
# Convert disclosed funding amounts to numeric values.
# Invalid or unavailable values are preserved as missing values.

df_final["Total_Funding_USD"] = pd.to_numeric(
    df_final["Total_Funding_USD"],
    errors="coerce"
)

print(df_final["Total_Funding_USD"].dtype)
```

```python
# Convert funding dates into a consistent datetime format.
# Invalid or unavailable dates become missing values.

df_final["Latest_Funding_Date"] = pd.to_datetime(
    df_final["Latest_Funding_Date"],
    errors="coerce"
)

print(df_final["Latest_Funding_Date"].dtype)
```

```python
# Confirm the final dataset dimensions before exporting.

print("Rows:", len(df_final))
print("Columns:", len(df_final.columns))
print("Column structure valid:", list(df_final.columns) == expected_columns)
```

## 18. Limitations

This research provides a structured view of the African AI companies identified within the defined research scope. It is not intended to represent a complete census of every AI company operating across Africa.

The main limitations include:

- Public information varies significantly between companies and countries.
- Some funding amounts and investor details are not publicly disclosed.
- Different sources may report different founding years, locations, or funding figures.
- Company information can change over time as businesses expand, relocate, raise new funding, or change their products.
- Some companies have limited publicly available information, resulting in partially verified records.
- Funding figures may represent different definitions of total funding depending on the source.

To maintain research integrity, unavailable information was left blank rather than estimated, and unresolved issues were documented in the research notes.
